### 👉 Conceptual Question:
Why do AI models struggle to understand sarcasm and context in text processing?
 1. **Figurative Language:** As a form of figurative language, sarcasm (like irony or hyperbole) is harder for models to process than direct, literal communication.

2.  **Absence of Tone:** Unlike spoken language, where prosody signals sarcasm, text-based models are deprived of this crucial cue, forcing them to rely on words that often express the opposite of the true meaning.


# **🧠 Discussion Questions**
1. How does AI process structured, image, and text data differently?
**Structured/Tabular Data:** This data is organized in rows and columns and is typically analyzed using machine learning or deep learning algorithms. The AI focuses on explicitly defined numerical or categorical features to identify patterns, correlations, and make predictions or classifications.
**Image Data:** Specialized deep learning models, most often Convolutional Neural Networks (CNNs), are used. The AI processes images pixel by pixel, learning hierarchical features (such as edges, shapes, and objects) to perform tasks like object detection, image classification, or segmentation.
**Text Data:** Processing involves Natural Language Processing (NLP) techniques and models like Transformers (e.g., GPT, BERT) or Recurrent Neural Networks (RNNs). A crucial first step is converting the text into numerical representations (embeddings) so the AI can learn the meaning, syntax, context, and semantic relationships within the language.

2. What real-world AI applications combine multiple types of data (structured, image, text)?
**Examples of Multimodal AI:**
**Self-Driving Cars:** These systems fuse various data types in real-time, including image and video from cameras, structured sensor data (like Lidar and radar for distance/speed), and text information from GPS and map databases.
**Medical Diagnosis:** AI assists doctors by analyzing image data (e.g., MRIs, X-rays), text data (e.g., patient history, notes), and structured data (e.g., blood test results, vital signs).
**Social Media Analysis/Recommendation Engines:** These tools generate content suggestions or perform sentiment analysis by integrating text (posts, comments), image/video, and structured data (e.g., user profiles, click-throughs, purchase history
